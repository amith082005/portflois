```
# Lambda decoder using dictionary
decode = lambda x: {
    1: "One",
    2: "Two",
    3: "Three"
}.get(x, "Unknown")  # Default to "Unknown" if key not found

# Test the decoder
print(decode(1))  # Output: One
print(decode(3))  # Output: Three
print(decode(5))  # Output: Unknown
```
