```
# Integer Example Program

# Declare integers
a = 10
b = 5

# Perform operations
sum_result = a + b
diff_result = a - b
product_result = a * b
quotient_result = a // b  # Integer division

# Display results
print("a =", a)
print("b =", b)
print("Sum =", sum_result)
print("Difference =", diff_result)
print("Product =", product_result)
print("Quotient =", quotient_result)
```
