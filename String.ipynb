```
# String Example Program

# Define a string
message = "Hello, Python World!"

# Print the string
print("Original Message:", message)

# String length
print("Length of message:", len(message))

# Accessing characters
print("First character:", message[0])
print("Last character:", message[-1])

# String slicing
print("First 5 characters:", message[:5])
print("Last 6 characters:", message[-6:])

# String methods
print("Uppercase:", message.upper())
print("Lowercase:", message.lower())
print("Is the message alphanumeric?", message.isalnum())

# Replacing a word
new_message = message.replace("World", "Programmer")
print("Updated Message:", new_message)

# Splitting the string
words = message.split(" ")
print("Words in message:", words)

# Loop through characters
print("Characters in message:")
for char in message:
    print(char, end=' ')
```
