```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Sample data (X = study hours, y = math score)
X = np.array([1, 2, 3, 4, 5, 6, 7]).reshape(-1, 1)
y = np.array([10, 40, 90, 160, 250, 360, 490])  # Non-linear to make log plot interesting

# Train model
model = LinearRegression()
model.fit(X, y)

# Predictions
y_pred = model.predict(X)

# Calculate Mean Squared Error
mse = mean_squared_error(y, y_pred)
print("Mean Squared Error (MSE):", mse)

# Plotting the actual and predicted values
plt.figure(figsize=(8, 5))
plt.plot(X, y, 'bo-', label='Actual Values')
plt.plot(X, y_pred, 'r--', label='Predicted Line')

# Logarithmic y-axis
plt.yscale('log')
plt.title("Actual vs Predicted (Log Scale)")
plt.xlabel("Study Hours")
plt.ylabel("Math Score (log scale)")
plt.legend()
plt.grid(True)
plt.show()
```
