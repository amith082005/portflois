```
# Import required libraries
from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data (Study Hours vs Math Scores)
# You can replace this with (Years vs Population) for population prediction
study_hours = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)   # Independent variable (X)
math_scores = np.array([50, 55, 65, 70, 75])            # Dependent variable (y)

# Create and train the model
model = LinearRegression()
model.fit(study_hours, math_scores)

# Predict score for 6 study hours
predicted_score = model.predict([[6]])

# Output
print("Predicted math score for 6 study hours:", predicted_score[0])
```
