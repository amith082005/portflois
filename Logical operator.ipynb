```
# Logical Operator Example Program

a = 10
b = 5
c = 15

# Using logical operators
print("a > b and a < c:", a > b and a < c)    # True and True => True
print("a > b or a > c:", a > b or a > c)      # True or False => True
print("not(a > b):", not(a > b))              # not(True) => False
```
