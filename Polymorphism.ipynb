```
# Parent class
class Animal:
    def speak(self):
        print("The animal makes a sound")

# Child class 1
class Dog(Animal):
    def speak(self):
        print("The dog barks")

# Child class 2
class Cat(Animal):
    def speak(self):
        print("The cat meows")

# Function that uses polymorphism
def make_animal_speak(animal):
    animal.speak()

# Create objects
a = Animal()
d = Dog()
c = Cat()

# Call function with different objects
make_animal_speak(a)
make_animal_speak(d)
make_animal_speak(c)
```
