```
from abc import ABC, abstractmethod

# Abstract class
class Person(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def display_role(self):
        pass  # Abstract method

# Child class 1
class Student(Person):
    def display_role(self):
        print(self.name, "is a Student.")

# Child class 2
class Teacher(Person):
    def display_role(self):
        print(self.name, "is a Teacher.")

# Create objects
s = Student("Amith")
t = Teacher("Ravi")

# Call methods
s.display_role()
t.display_role()
```
